```python
# read in batches of 1000 with pandas
chunksize = 100000
accessions = []
for chunk in pd.read_csv("data/train_shuffled.csv", chunksize=chunksize):
    # chunk is a DataFrame. To "process" the rows in the chunk:
    accessions.extend(chunk['accession'].values)
```

```python
swiss_prot_plants = swiss_prot_plants[~swiss_prot_plants['accession'].isin(accessions)]
swiss_prot_plants
```

```python
validation = pd.read_csv("data/validation.csv")
```

```python
swiss_prot_plants = swiss_prot_plants[~swiss_prot_plants['accession'].isin(validation['accession'].values)]
swiss_prot_plants
```

```python
import re

def get_ec_from_regex_match(match):
    if match is not None:
        EC = match.group()
        if EC is not None:
            return EC
    return None

def get_ecs_from_string(EC):
    EC3 = []
    EC2 = []
    EC1 = []
    EC4 = []
    new_EC = re.search(r"^\d+.\d+.\d+.n*\d+", EC)
    new_EC = get_ec_from_regex_match(new_EC)
    if isinstance(new_EC, str):
        if new_EC not in EC4:
            EC4.append(new_EC)

    new_EC = re.search(r"^\d+.\d+.\d+", EC)
    new_EC = get_ec_from_regex_match(new_EC)
    if isinstance(new_EC, str):
        if new_EC not in EC3:
            EC3.append(new_EC)

    new_EC = re.search(r"^\d+.\d+", EC)
    new_EC = get_ec_from_regex_match(new_EC)
    if isinstance(new_EC, str):
        if new_EC not in EC2:
            EC2.append(new_EC)

    new_EC = re.search(r"^\d+", EC)
    new_EC = get_ec_from_regex_match(new_EC)
    if isinstance(new_EC, str):
        if new_EC not in EC1:
            EC1.append(new_EC)
    return EC1, EC2, EC3, EC4
```

```python
validation.iloc[:, 8:]
# create new columns for each EC in the swiss_prot_plants dataset
for EC in validation.iloc[:, 8:]:
    
    swiss_prot_plants[EC] = 0
    
```

```python
columns = swiss_prot_plants.columns
for i, row in swiss_prot_plants.iterrows():
    EC1, EC2, EC3, EC4 = get_ecs_from_string(row['EC'])
    for EC in EC1:
        if EC in columns:
            swiss_prot_plants.at[i, EC] = 1
    for EC in EC2:
        if EC in columns:
            swiss_prot_plants.at[i, EC] = 1
    for EC in EC3:
        if EC in columns:
            swiss_prot_plants.at[i, EC] = 1
    for EC in EC4:
        if EC in columns:
            swiss_prot_plants.at[i, EC] = 1
```

```python
swiss_prot_plants.to_csv('swiss_prot_ec_plants.csv', index=False)
```

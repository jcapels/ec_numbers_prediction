```python
def data_by_record(filename):
    """
    Auxiliar function to read the dat file and divide the info by record
    Parameters
    ----------
    filename: str
    json_path for the dat file

    Returns
    -------
    all_records: list
    list of dicts where each dict has the info of each record
    """
    all_records = []
    datafile = open(filename, 'r', encoding='ISO-8859-1')
    line = datafile.readline().strip()

    while line:
        if line[0] != "#" and "UNIQUE-ID" in line:
            record = []
            while line != '//':
                record.append(line)
                line = datafile.readline().strip()

            record_dic = {}
            for info in record:
                if info.count(' - ') == 1:
                    key, value = info.split(' - ')
                    if key not in record_dic:
                        record_dic[key] = [value]
                    else:
                        record_dic[key].append(value)

            all_records.append(record_dic)
        else:
            line = datafile.readline().strip()

    return all_records
```

```python
from anytree import Node, RenderTree

records = data_by_record('data/classes.dat')
pathways = ["SECONDARY-METABOLITE-BIOSYNTHESIS"]
nodes = {"SECONDARY-METABOLITE-BIOSYNTHESIS": Node("SECONDARY-METABOLITE-BIOSYNTHESIS")}
for record in records:
    types_ = record['TYPES']
    for type_ in types_:
        if type_ in pathways:
            nodes[record['UNIQUE-ID'][0]] = Node(record['UNIQUE-ID'][0], parent=nodes[type_], pathway = record['UNIQUE-ID'][0])
            pathways.extend(record['UNIQUE-ID'])
```

```python
print(RenderTree(nodes["SECONDARY-METABOLITE-BIOSYNTHESIS"]))
```

```python
secondary_metabolites_pathways_classes = []
for node in nodes["SECONDARY-METABOLITE-BIOSYNTHESIS"].leaves:
    secondary_metabolites_pathways_classes.append(node.pathway)
```

```python
# export list to pickle
import pickle

with open('data/secondary_metabolites_pathway_classes.pkl', 'wb') as f:
    pickle.dump(secondary_metabolites_pathways_classes, f)
```

```python
# import list from pickle
import pickle

with open('data/secondary_metabolites_pathway_classes.pkl', 'rb') as f:
    secondary_metabolites_pathways_classes = pickle.load(f)
```

```python
pathways = data_by_record("data/pathways.dat")
secondary_metabolites_pathways = []
for pathway in pathways:
    if pathway["TYPES"][0] in secondary_metabolites_pathways_classes:
        secondary_metabolites_pathways.append(pathway["UNIQUE-ID"][0])
```

```python
with open('data/secondary_metabolites_pathway.pkl', 'wb') as f:
    pickle.dump(secondary_metabolites_pathways, f)
```

```python
with open('data/secondary_metabolites_pathway.pkl', 'rb') as f:
    secondary_metabolites_pathways = pickle.load(f)
```

```python
reactions = data_by_record("data/reactions.dat")
secondary_metabolites_reactions = []
secondary_metabolism_enzymes = {}
for reaction in reactions:
    if "IN-PATHWAY" in reaction:
        pathways = reaction["IN-PATHWAY"]
        for pathway in pathways:
            if pathway in secondary_metabolites_pathways:
                secondary_metabolites_reactions.append(reaction["UNIQUE-ID"][0])
                if "EC-NUMBER" in reaction:
                    ec_numbers = reaction["EC-NUMBER"]
                    for ec_number in ec_numbers:
                        if pathway not in secondary_metabolism_enzymes:
                            secondary_metabolism_enzymes[pathway] = [ec_number.replace("EC-","")]
                        else:
                            secondary_metabolism_enzymes[pathway].append(ec_number.replace("EC-",""))
```

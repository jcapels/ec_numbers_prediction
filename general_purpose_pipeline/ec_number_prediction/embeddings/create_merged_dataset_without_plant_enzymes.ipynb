```python
# read pickle file
import pickle
import os

datasets = ["train", "validation", "test"]

current_dataset = {"place_holder": {}}
for dataset in datasets:

    with open(f'esm1b/{dataset}/features.pkl', 'rb') as f:
        esm1b_dataset = pickle.load(f)
    
    current_dataset["place_holder"].update(esm1b_dataset["place_holder"])

os.makedirs("esm1b/merged/", exist_ok=True)
with open(f'esm1b/merged/features.pkl', 'wb') as f:
    pickle.dump(current_dataset, f)
```

```python
import pandas as pd

dataset = pd.read_csv("merged_dataset.csv")
```

```python
dataset.shape
```

```python
plants = pd.read_csv("swiss_prot_ec_plants.csv")
```

```python
new_dataset = dataset[~dataset["accession"].isin(plants["accession"])]
new_dataset.head()
```

```python
# check the the columns that have only zeros
zero_columns = new_dataset.iloc[:, 8:].sum() == 0
```

```python
ecs_to_delete = []
for i, column in enumerate(zero_columns):
    if column:
        ecs_to_delete.append(zero_columns.index[i])
```

```python
new_dataset.drop(ecs_to_delete, axis=1, inplace=True)
```

```python
new_dataset.shape
```

```python
new_dataset.to_csv("merged_dataset_no_plants.csv", index=False)
```

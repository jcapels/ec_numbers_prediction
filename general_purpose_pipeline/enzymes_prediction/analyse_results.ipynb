```python
import pandas as pd

results = pd.read_csv('results_esm1b_enzyme.csv')
```

```python
results["Model"] = "ESM1b"
```

```python
results
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Creating a boxplot
plt.figure(figsize=(12, 6))
sns.boxplot(x='Model', y='accuracy_score', hue='Model type', data=results)
plt.title('Accuracy Score by Model (TF and with no TF)') 
plt.xlabel('Model')
plt.ylabel('Accuracy Score')
plt.xticks(rotation=45)  # Rotate model names for better readability
plt.tight_layout()
plt.show()

```

```python
# wilcoxon test
import scipy.stats as stats

# ESM-2 3B
stats.wilcoxon(results[(results["Model"] == "ESM1b") & (results["Model type"] == "TF")]["accuracy_score"],
               results[(results["Model"] == "ESM1b") & (results["Model type"] == "No TF")]["accuracy_score"])
```

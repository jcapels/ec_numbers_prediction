```python
import pandas as pd

dataset = pd.read_csv("swiss_prot_enzymes_filtered.csv")

dataset 
```

```python
# Export as FASTA file
with open('swiss_prot_enzymes_filtered.fasta', 'w') as f:
    for idx, row in dataset.iterrows():
        f.write(f">{row['accession']}\n{row['sequence']}\n")
```

```python
# shuffle the dataset
dataset = dataset.sample(frac=1).reset_index(drop=True)
dataset
```

```python
# split the data with sklearn 
from sklearn.model_selection import train_test_split

train, test = train_test_split(dataset, test_size=0.05)
print(test.shape)
train, validation = train_test_split(train, test_size=0.05)
print(train.shape)
print(validation.shape)
train.to_csv("./data/train.csv", index=False)
validation.to_csv("./data/validation.csv", index=False)
test.to_csv("./data/test.csv", index=False)
```

```python
# count the number of classes
print(train["enzyme"].value_counts())
print(validation["enzyme"].value_counts())
print(test["enzyme"].value_counts())
```
